**Data Cleaning and Preprocessing**

```python
df_cleaned = df.copy()
```

```python
df_cleaned.isna().sum()
```

```python
df_cleaned.shape
```

```python
df_cleaned.drop_duplicates(inplace=True)
```

```python
df_cleaned.shape
```

```python
df_cleaned.dtypes
```

```python
df_cleaned = pd.get_dummies(df_cleaned, columns=['gender', 'course', 'sleep_quality', 'study_method', 'facility_rating', 'exam_difficulty'])
```

```python
df_cleaned
```

```python
df_cleaned = df_cleaned.replace({True:1, False:0})
```

```python
df_cleaned['internet_access'] = df_cleaned['internet_access'].map({"yes" : 1, "no" : 0})
```

```python
df_cleaned
```

```python
df_cleaned['study_hours'] = pd.cut(
    df_cleaned['study_hours'],
    bins = [0.08, 1.68,3.28, 4.88,6.48, float('inf')],
    labels = ["Very Low", "Low", "Moderate", "High", "Very High"]
)
```

```python
df_cleaned
```

```python
df_cleaned = pd.get_dummies(df_cleaned, columns=['study_hours'])
```

```python
df_cleaned
```

```python
df_cleaned = df_cleaned.replace({True:1, False:0})
```

```python
df_cleaned['class_attendance'] = pd.cut(
    df_cleaned['class_attendance'],
    bins = [40.60, 52.60,64.60, 76.60,88.60, float('inf')],
    labels = ["Very Poor", "Poor", "Average", "Gpod", "Excellent"]
)
```

```python
df_cleaned = pd.get_dummies(df_cleaned, columns=['class_attendance'])
```

```python
df_cleaned = df_cleaned.replace({True:1, False:0})
```

```python
df_cleaned.head()
```

```python
df_cleaned.drop(columns=['student_id', 'age'], inplace=True)
```

```python
df_cleaned['sleep_hours'] = pd.cut(
    df_cleaned['sleep_hours'],
    bins = [4.10, 5.30, 6.50, 7.70, 8.90, float('inf')],
    labels = ["Very Low Sleep", "Low Sleep", "Normal Sleep", "Good Sleep", "Excellent Sleep"]
)
```

```python
df_cleaned = pd.get_dummies(df_cleaned, columns=['sleep_hours'])
```

```python
df_cleaned.replace({True:1, False:0}, inplace=True)
```
